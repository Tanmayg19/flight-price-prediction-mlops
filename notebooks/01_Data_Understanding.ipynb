## 1.4 Dataset Relationships and Data Validation

```python
# Check unique users across datasets

print("Users in Users dataset   :", users["code"].nunique())
print("Users in Flights dataset :", flights["userCode"].nunique())
print("Users in Hotels dataset  :", hotels["userCode"].nunique())
```

```python
# Checking whether all flight users exist in Users dataset

flight_users_not_in_users = set(flights["userCode"]) - set(users["code"])

print("Flight users not found in Users dataset:",
      len(flight_users_not_in_users))
```

```python
# Checking whether all hotel users exist in Users dataset

hotel_users_not_in_users = set(hotels["userCode"]) - set(users["code"])

print("Hotel users not found in Users dataset:",
      len(hotel_users_not_in_users))
```

```python
# Checking travelCode relationship

flight_travel_codes = set(flights["travelCode"])
hotel_travel_codes = set(hotels["travelCode"])

print("Travel codes in Flights but not Hotels:",
      len(flight_travel_codes - hotel_travel_codes))

print("Travel codes in Hotels but not Flights:",
      len(hotel_travel_codes - flight_travel_codes))
```
